# 17. Hipótesis 1: 
¿Los hijos aseguran el viaje? Escribe una consulta usando CASE WHEN que clasifique a la reserva como "Con Hijos" (si children > 0 o babies > 0) o "Sin Hijos". Agrupa por esta nueva categoría y calcula la tasa de cancelación (puedes sacar el promedio - AVG - de la columna is_canceled). ¿Qué segmento cancela más?

```python
%sql
SELECT 
    CASE 
        WHEN children>0 OR babies>0 THEN 'Con Hijos'    
        ELSE 'Sin Hijos'
    END AS Clasificacion_reserva,
    round(Avg(is_canceled),4) as Tasa_Cancelacion
FROM hotel_dataandino.proyecto_hotel_db.tb_reservas
GROUP BY clasificacion_reserva;
```

### Interpretación
- Se identificó que las reservas sin hijos presentan una tasa de cancelación del 37.67 %, mientras que las reservas con hijos registran un 35.24 %. Esto indica que las reservas sin hijos tienen una mayor proporción de cancelaciones.
- La diferencia es de aproximadamente 2.43 % puntos. Por lo tanto, viajar con hijos no asegura que la reserva se mantenga, pero en estos datos las reservas con hijos presentan una menor tasa de cancelación.

# 18. Hipótesis 2: El impacto de las solicitudes especiales. 
Agrupa por la columna total_of_special_requests y calcula la tasa de cancelación (AVG(is_canceled)) para cada cantidad de solicitudes (0, 1, 2, etc.). Ordena por la cantidad de solicitudes de forma ascendente. ¿A medida que el cliente pide más cosas, cancela más o cancela menos?

```python
%sql
SELECT 
  total_of_special_requests AS Cantidad_solicitudes_especiales,
   round(Avg(is_canceled),4) as Tasa_Cancelacion
FROM hotel_dataandino.proyecto_hotel_db.tb_reservas
GROUP BY total_of_special_requests
ORDER BY total_of_special_requests ASC;
```

### Interpretación
- Se identificó que los clientes que realizan más solicitudes especiales presentan una menor tasa de cancelación. Esta llega al 5%, en comparación con el 48.31% de las reservas sin solicitudes especiales.

# 19. Hipótesis 3: ¿Cambiar la reserva es señal de compromiso? 
Al igual que en los ejercicios anteriores, utiliza CASE WHEN para segmentar la columna booking_changes en tres grupos: "Sin Cambios" (0), "1 a 3 cambios" y "Más de 3 cambios". Calcula la tasa de cancelación para cada grupo. ¿Quiénes tienen menor riesgo de cancelar?

```python
%sql
SELECT
    CASE
        WHEN booking_changes = 0 THEN 'Sin Cambios'
        WHEN booking_changes BETWEEN 1 AND 3 THEN '1 a 3 cambios'
        ELSE 'Más de 3 cambios' 
    END AS Cantidad_cambios,
    round(Avg(is_canceled),4) as Tasa_Cancelacion
FROM hotel_dataandino.proyecto_hotel_db.tb_reservas
GROUP BY Cantidad_cambios
ORDER BY Tasa_Cancelacion DESC;
```

### Interpretación
- Se identificó que las reservas con 1 a 3 cambios presentan la menor tasa de cancelación 15.89%. En cambio, las reservas sin cambios tienen la tasa más alta 41.24%.

# 20. Hipótesis 4: El factor precio (adr). 
Agrupa las reservas por tipo de hotel (hotel). Para cada tipo, muestra la cantidad total de cancelaciones y el precio promedio (adr) solo de las reservas canceladas. En base a tu consulta, ¿en qué tipo de hotel el precio promedio de cancelación es más alto?

```python
%sql
SELECT 
    hotel,
    count(is_canceled) as Total_Cancelaciones,
    round(avg(adr),2)as Precio_Promedio_Cancelaciones
FROM hotel_dataandino.proyecto_hotel_db.tb_reservas
WHERE is_canceled=1
GROUP BY hotel
ORDER BY Precio_Promedio_Cancelaciones DESC
```

### Interpretación
- Se identificó que el Resort Hotel presenta un precio promedio de cancelación ligeramente mayor 106.59, que el City Hotel 105.10. Sin embargo, el City Hotel registra más cancelaciones.
