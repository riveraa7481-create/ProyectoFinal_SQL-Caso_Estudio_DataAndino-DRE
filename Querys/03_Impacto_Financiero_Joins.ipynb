# 11. El Costo del Marketing Inútil (Agregación con Filtro):
Sabiendo que a la agencia
(canal TA/TO) se le paga USD 1.5 por cada reserva. ¿Cuánto dinero en total se ha
pagado a la agencia por reservas que finalmente fueron canceladas (is_canceled =
1)? Escribe la consulta que calcule este monto de pérdida.

```python
%sql
SELECT 
    distribution_channel,
    is_canceled,
    COUNT(is_canceled) * 1.5 AS Monto_perdida
FROM hotel_dataandino.proyecto_hotel_db.tb_reservas
WHERE distribution_channel = 'TA/TO' AND  is_canceled = 1	
GROUP BY distribution_channel,is_canceled
```

### Interpretación del resultado:
- Se identificó que el canal TA/TO generó un monto de USD 60,097.50 por reservas que finalmente fueron canceladas. Este monto representa el pago realizado a la agencia por reservas que no llegaron a concretarse.

# 12. El Costo de Cancelar Tarde (DATE_DIFF): 
(Considera que tienes las columnas
arrival_date y reservation_status_date ambas en formato DATE). Cuando un
huésped cancela con menos de 3 días de anticipación respecto a su llegada, el hotel
pierde USD 120. Escribe una consulta que cuente cuántas reservas cumplen esta
condición (is_canceled = 1 y diferencia entre llegada y cancelación menor a 3 días)
y multiplica esa cantidad por 120 para ver la pérdida total.

```python
%sql
WITH reservas AS (
    SELECT
        is_canceled,
        TO_DATE(
            CONCAT(
                CAST(arrival_date_day_of_month AS STRING), '-',
                arrival_date_month, '-',
                CAST(arrival_date_year AS STRING)
            ),
            'd-MMMM-yyyy'
        ) AS arrival_date,
        TO_DATE(reservation_status_date, 'd/M/yy') AS reservation_status_date
    FROM hotel_dataandino.proyecto_hotel_db.tb_reservas
)
SELECT
    COUNT(*) AS reservas_canceladas_tarde,
    COUNT(*) * 120 AS perdida_total
FROM reservas
WHERE is_canceled = 1 
AND DATEDIFF(arrival_date, reservation_status_date) < 3;

```

### Interpretación del resultado:
Se identificaron 3,085 reservas canceladas con menos de 3 días de anticipación respecto a la fecha de llegada. Considerando un costo de USD 120 por cada cancelación tardía, se obtuvo una pérdida total estimada de USD 370,200 para el hotel.

# 14. (JOIN 2 - INNER JOIN con Agrupación): 
Usando el mismo JOIN del ejercicio anterior, averigua: ¿Cuáles son los 5 países (muestra el nombre_pais) que tienen la mayor cantidad de reservas canceladas?

```python
%sql
SELECT
    P.nombre_pais AS Pais,
    R.is_canceled AS Reserva_Cancelada,
    count(R.is_canceled) AS Cantidad_Canceladas
FROM hotel_dataandino.proyecto_hotel_db.tb_reservas R 
INNER JOIN hotel_dataandino.proyecto_hotel_db.tb_paises P
ON R.country = P.cod_pais
WHERE R.is_canceled=1
GROUP BY R.is_canceled, P.nombre_pais
ORDER BY Cantidad_Canceladas DESC
LIMIT 5;
```

### Interpretación del resultado:
- Se relacionaron las reservas con la tabla de países mediante un INNER JOIN. Luego se contaron las reservas canceladas por país y se ordenaron de mayor a menor. Finalmente, se identificaron los cinco países con la mayor cantidad de cancelaciones.

# Creación del identificador único de reservas.

```python
%sql
CREATE OR REPLACE TABLE hotel_dataandino.proyecto_hotel_db.tb_reservas_prueba AS
SELECT *,
    ROW_NUMBER() OVER (
        ORDER BY arrival_date_year, arrival_date_day_of_month
    ) AS id_reserva
FROM hotel_dataandino.proyecto_hotel_db.tb_reservas;
```


### Interpretación:
- Se creó la tabla tb_reservas_prueba como una copia de tb_reservas, agregando la columna id_reserva mediante ROW_NUMBER() para generar un identificador secuencial para cada reserva.

```python
%sql
ALTER TABLE hotel_dataandino.proyecto_hotel_db.tb_reservas_prueba
ADD COLUMNS (id_vendedor INT);
```

### Interpretación
- Se agregó una nueva columna llamada id_vendedor de tipo entero (INT) a la tabla tb_reservas_prueba. Esta modificación permitirá asignar un vendedor a cada reserva y establecer posteriormente una relación entre ambas tablas mediante el identificador del vendedor.

```python
%sql
UPDATE hotel_dataandino.proyecto_hotel_db.tb_reservas_prueba
SET id_vendedor = CAST(((id_reserva - 1) % 20) + 1 AS INT);
```

### Interpretación
- Se asignó un identificador de vendedor a cada reserva de la tabla tb_reservas_prueba, distribuyendo de manera repetitiva los 20 vendedores disponibles. De esta manera, cada reserva quedó asociada a un vendedor, permitiendo establecer posteriormente una relación entre ambas tablas mediante id_vendedor.

### 15. (JOIN 3 - Simulación FULL OUTER JOIN): 
Supongamos que Recursos Humanos tiene una tabla tb_vendedores (con id_vendedor y nombre_vendedor) y cruzamos las reservas para saber qué vendedor atendió qué reserva. Realiza un FULL OUTER JOIN entre las reservas y los vendedores. Selecciona el canal de distribución y el nombre del vendedor para identificar si hay reservas sin vendedor o vendedores sin reservas asignadas.

```python
%sql
select
    R.distribution_channel,
    V.nombre_vendedor,
    R.id_reserva
from hotel_dataandino.proyecto_hotel_db.tb_reservas_prueba R
full outer join hotel_dataandino.proyecto_hotel_db.tb_vendedores V
on R.id_vendedor= V.id_vendedor

```

### Interpretación del resultado:
Se relacionaron las reservas con la tabla de vendedores mediante un FULL OUTER JOIN, comparando id_vendedor con id_vendedor. Se obtuvo el canal de distribución y el nombre del vendedor, permitiendo identificar los registros que tienen coincidencias y aquellos que no encuentran relación, los cuales aparecen como NULL. 

# 16. (JOIN 4 - RIGHT JOIN): 
El departamento de contabilidad tiene una tabla de facturación (tb_facturas con id_reserva y monto_facturado). Usando un RIGHT JOIN desde las reservas hacia las facturas, garantiza que veamos todas las facturas emitidas, cruzadas con el tipo de hotel (hotel).

```python
%sql
SELECT 
    R.hotel,
    F.monto_facturado
FROM hotel_dataandino.proyecto_hotel_db.tb_reservas_prueba R
RIGHT JOIN hotel_dataandino.proyecto_hotel_db.tb_facturas F
ON R.id_reserva= F.id_reserva
ORDER BY F.monto_facturado DESC;

```

### Interpretación del resultado:
- Se relacionaron las tablas tb_reservas_prueba y tb_facturas mediante un RIGHT JOIN para visualizar todas las facturas emitidas junto con el tipo de hotel correspondiente. Además, se ordenaron los montos facturados de mayor a menor.
